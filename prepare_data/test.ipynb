```python
import pandas as pd

# 1. Wir laden das UNGEFILTERTE Original-Ergebnis (wo alle 500 drin sind)
# (Passe den Dateinamen an, wenn deine Datei mit den 500 rohen Zeilen anders heißt)
df_raw = pd.read_csv("ergebnis.csv")

# 2. DataFrame NUR auf Baden-Württemberg einschränken
df_bw = df_raw[df_raw['source_name'] == 'source.court.baden-württemberg']

# 3. Zählen, wie viele BW-Dokumente es insgesamt gibt
gesamtzahl_bw = len(df_bw)
print(f"Baden-Württemberg Datensätze insgesamt: {gesamtzahl_bw}")

if gesamtzahl_bw > 0:
    # 4. Prüfen, wie viele davon das Urteils-Muster (z.B. NJRE + Zahlen) NICHT haben.
    # ~ (Tilde) negiert die Bedingung "enthält NJRE oder JURE etc.".
    # Falls das 'RE' Muster immer gilt:
    muster = r"[A-Z]*RE\d+"
    
    mask_hat_muster = df_bw['url_url'].str.contains(muster, na=False, regex=True)
    mask_hat_kein_muster = ~mask_hat_muster
    
    df_bw_ohne_re = df_bw[mask_hat_kein_muster]
    
    print(f"Davon ERFÜLLEN NICHT das Urteils-Muster: {len(df_bw_ohne_re)}")
    print(f"Davon SIND vermutlich echte Urteile: {mask_hat_muster.sum()}")
    
    # 5. Lass uns zur Kontrolle einen Blick auf die URLs werfen, die NICHT das Muster haben
    if not df_bw_ohne_re.empty:
        print("\n--- Beispiel-URLs der als NICHT-Urteile erkannten Dokumente ---")
        for url in df_bw_ohne_re['url_url'].head(10):
            print(url)
else:
    print("Keine Datensätze aus Baden-Württemberg gefunden. Bitte rohes 'ergebnis.csv' überprüfen.")
```
